```python
import pandas as pd

df = pd.DataFrame({'Date': ['2023-01-01', '2023-01-02', '2023-01-03'],
                   'Sales': [100, 120, 130]})

df['7_day_avg'] = df['Sales'].rolling(window=7).mean()
print(df)
```

```python
# Here we are specifying the average window as 7, so the mean is calculated for 
# seven days and the value is displayed after 7 days..(i.e mean calculated for 1 to 7 days and displayed on the 7th day)
# here we have only data for three days so it got displayed as Nan
```
