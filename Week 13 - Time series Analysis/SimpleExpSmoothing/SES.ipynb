```python
import pandas as pd
```

```python
stk_data=pd.read_csv("TATACOFEE_1321.csv")
```

```python
stk_data.rename(columns={"Price":"Close"},inplace=True)
```

```python
import warnings
warnings.filterwarnings("ignore")
```

```python
# Filtering data based on the required dates
stk_data["Date"]=pd.to_datetime(stk_data["Date"])
stk_data=stk_data[(stk_data["Date"]>="2020-07-01")&(stk_data["Date"]<="2021-12-31")]
```

```python
# Preprocessing 
from sklearn.preprocessing import MinMaxScaler
Ms= MinMaxScaler()
data1=Ms.fit_transform(stk_data[["Close"]])
print("Len:",data1.shape)
```

```python
import warnings 
warnings.filterwarnings("ignore")
```

```python
from statsmodels.tsa.holtwinters import SimpleExpSmoothing
from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_percentage_error
```

```python

# Number of observations for testing
test_obs = 28

# Split data into training and testing
train = data1[:-test_obs]
test = data1[-test_obs:]

# Build and fit SES model
model = SimpleExpSmoothing(train, initialization_method="estimated")
result = model.fit()

# Forecast the next 28 observations
y_pred = result.forecast(test_obs)

# RMSE
rmse = round(mean_squared_error(test, y_pred)**0.5, 4)

# MAPE
mape = mean_absolute_percentage_error(test, y_pred)

print("RMSE:", rmse)
print("MAPE:", mape)
```
