```python
import pandas as pd

df = pd.DataFrame({'Date': ['2023-01-01', '2023-01-02', '2023-01-03'],
                   'Sales': [100, 120, 130]})

df['Sales'].isnulls().sum()
```

```python
# Correct code
import pandas as pd

df = pd.DataFrame({'Date': ['2023-01-01', '2023-01-02', '2023-01-03'],
                   'Sales': [100, 120, 130]})

df['Sales'].isnull().sum()
# it returns 0 because there is no null value
```

# Q2.TypeError

```python
df.resample('M').mean()
#TypeError: Only valid with DatetimeIndex, TimedeltaIndex or PeriodIndex, but got an instance of 'RangeIndex'
```

```python
# Here we are setting the datetime index to group the data on monthly basis
df['Date'] = pd.to_datetime(df['Date'])
df.set_index('Date', inplace=True)
df.resample('M').mean()
```

# Q3.ValueError

```python
import pandas as pd
import numpy as np
data = {
    'Date': pd.date_range(start='2023-01-01', periods=10, freq='D'),
    'Passengers': [100, 120, np.nan, 140, 135, np.nan, 150, 160, 170, np.nan]
}

df = pd.DataFrame(data)
print("Original Data:\n", df)
df['Passengers'] = df['Passengers'].astype(int)
#intcastingNanError, pandas cannot convert Nan values to integer
```

```python
#correct code
import pandas as pd
import numpy as np
data = {
    'Date': pd.date_range(start='2023-01-01', periods=10, freq='D'),
    'Passengers': [100, 120, np.nan, 140, 135, np.nan, 150, 160, 170, np.nan]
}
df = pd.DataFrame(data)
# so here we are replacing the na values with mean value 
df["Passengers"]=df["Passengers"].fillna(df["Passengers"].mean())
print("Original Data:\n", df)
df['Passengers'] = df['Passengers'].astype(int)
```
